- Create a base class `Shape` with a method `area()` that raises `NotImplementedError`.
- Create subclasses `Rectangle` and `Circle` that implement `area()` appropriately.

````python
import math


class Shape:

    def area(self):
        raise NotImplementedError


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius


rectangle = Rectangle(4, 2)
circle = Circle(9)

print("Rectangle area:", rectangle.area())
print("Circle area:", circle.area())
````

- Create a base class `BankAccount` with `owner`, `balance`, and methods `deposit()` and `withdraw()`.
- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.
- Demonstrate deposits, withdrawals, and interest application.

````python
class BankAccount:

    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")


class SavingsAccount(BankAccount):

    def __init__(self, owner, balance, interest_rate):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.balance * self.interest_rate / 100
        self.balance = self.balance + interest
        print("Interest added:", interest)


account = SavingsAccount("Frank", 10000, 15)

print("Owner:", account.owner)
print("Starting balance:", account.balance)

account.deposit(1200)
account.withdraw(700)
account.add_interest()

print("Final balance:", account.balance)
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:

    def __init__(self, n):
        self.n = n

    def __iter__(self):
        return self

    def __next__(self):

        if self.n <= 0:
            raise StopIteration

        number = self.n
        self.n = self.n - 1

        return number


count = CountDown(5)

for number in count:
    print(number)
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
with open("students.txt", "w") as file:
    file.write("Frank\n")
    file.write("John\n")
    file.write("Mary\n")
    file.write("Peter\n")
````

````python
class FileLineReader:

    def __init__(self, filename):
        self.file = open(filename, "r")

    def __iter__(self):
        return self

    def __next__(self):

        line = self.file.readline()

        if line == "":
            self.file.close()
            raise StopIteration

        return line.strip()


reader = FileLineReader("students.txt")

for line in reader:
    print(line)
````

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
from abc import ABC, abstractmethod
import math


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius


shapes = [
    Rectangle(12, 4),
    Circle(9),
    Rectangle(7, 3)
]

for shape in shapes:
    print("Area:", shape.area())
````

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:

    def __init__(self, n):
        self.n = n
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):

        if self.count >= self.n:
            raise StopIteration

        number = self.a

        self.a = self.b
        self.b = self.a + self.b

        self.count = self.count + 1

        return number


fib = Fibonacci(10)

for number in fib:
    print(number)
````

# Library Management System  
  
## Scenario  
You are building a small library system to manage books, members, and borrowing operations. The system should support multiple types of books, track borrowed books, and allow iteration over members and books. The system should also demonstrate proper use of scopes and encapsulation.  
  
## Requirements and Concepts to Cover  
  
### 1. OOP & Inheritance  
- Create a base class `Book` with attributes: `title`, `author`, `year`.  
- Create subclasses:  
  - `EBook` with `file_size` and `format`.  
  - `PrintedBook` with `weight` and `dimensions`.  
- Create a base class `Member` with `name` and `member_id`.  
- Create subclasses:  
  - `StudentMember` with `grade_level`.  
  - `TeacherMember` with `department`.  
  
### 2. Polymorphism  
- Define a method `get_description()` in `Book` and override it in both `EBook` and `PrintedBook`.  
- Define a method `borrow_book()` in `Member` that behaves differently for `StudentMember` and `TeacherMember` (e.g., different borrowing limits).  
  
### 3. Iterators  
- Implement an iterator for the `Library` class to iterate over all books.  
- Implement an iterator for the `Member` class to iterate over borrowed books.  
  
### 4. Scope  
- Use private and protected attributes to demonstrate scope:  
  - `_borrowed_books` in `Member` (protected).  
  - `__library_name` in `Library` (private).  
- Provide getter and setter methods where appropriate to access private/protected attributes.  
  
### 5. Extra Features  
- Track which books are available and which are borrowed.  
- Allow searching books by title or author.  
- Count the total number of books borrowed by all members (use a class-level variable).  

````python
class Book:

    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.is_borrowed = False

    def get_description(self):
        return f"{self.title} by {self.author} ({self.year})"


class EBook(Book):

    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    def get_description(self):
        return f"EBook: {self.title} by {self.author}, {self.file_size}MB, {self.format}"


class PrintedBook(Book):

    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return f"Printed Book: {self.title} by {self.author}, {self.weight}kg, {self.dimensions}"


class Member:

    total_borrowed = 0

    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self._borrowed_books = []

    def borrow_book(self, book):
        print("Member borrowing book")

    def return_book(self, book):
        if book in self._borrowed_books:
            self._borrowed_books.remove(book)
            book.is_borrowed = False
            Member.total_borrowed = Member.total_borrowed - 1
            print(book.title, "has been returned.")
        else:
            print("This book was not borrowed by this member.")

    def __iter__(self):
        return iter(self._borrowed_books)

    def get_borrowed_books(self):
        return self._borrowed_books


class StudentMember(Member):

    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id)
        self.grade_level = grade_level
        self.borrowing_limit = 3

    def borrow_book(self, book):

        if len(self._borrowed_books) >= self.borrowing_limit:
            print("Student has reached the borrowing limit.")
            return

        if book.is_borrowed:
            print("This book is already borrowed.")
            return

        self._borrowed_books.append(book)
        book.is_borrowed = True
        Member.total_borrowed = Member.total_borrowed + 1

        print(self.name, "borrowed", book.title)


class TeacherMember(Member):

    def __init__(self, name, member_id, department):
        super().__init__(name, member_id)
        self.department = department
        self.borrowing_limit = 5

    def borrow_book(self, book):

        if len(self._borrowed_books) >= self.borrowing_limit:
            print("Teacher has reached the borrowing limit.")
            return

        if book.is_borrowed:
            print("This book is already borrowed.")
            return

        self._borrowed_books.append(book)
        book.is_borrowed = True
        Member.total_borrowed = Member.total_borrowed + 1

        print(self.name, "borrowed", book.title)


class Library:

    def __init__(self, library_name):
        self.__library_name = library_name
        self.books = []
        self.members = []

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def get_library_name(self):
        return self.__library_name

    def set_library_name(self, new_name):
        self.__library_name = new_name

    def search_books(self, search_term):

        results = []

        for book in self.books:

            if (search_term.lower() in book.title.lower()
                    or search_term.lower() in book.author.lower()):
                results.append(book)

        return results

    def __iter__(self):
        return iter(self.books)



# Create books


book1 = EBook(
    "Python Basics",
    "John Smith",
    2024,
    5,
    "PDF"
)

book2 = PrintedBook(
    "Introduction to AI",
    "Mary Jones",
    2023,
    1.2,
    "20 x 15 cm"
)

book3 = PrintedBook(
    "Data Science Guide",
    "John Smith",
    2025,
    1.5,
    "25 x 18 cm"
)



# Create members


student = StudentMember(
    "Frank",
    "ST001",
    "Level 200"
)

teacher = TeacherMember(
    "Mr. Mensah",
    "TC001",
    "Computer Science"
)



# Create library


library = Library("Central Library")

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

library.add_member(student)
library.add_member(teacher)



# Polymorphism


print("BOOK DESCRIPTIONS")
print("------------------")

for book in library:
    print(book.get_description())



# Borrow books


print("\nBORROWING BOOKS")
print("----------------")

student.borrow_book(book1)
student.borrow_book(book2)

teacher.borrow_book(book3)



# Iterate over borrowed books


print("\nSTUDENT BORROWED BOOKS")
print("----------------------")

for book in student:
    print(book.title)



# Search books


print("\nSEARCH RESULTS")
print("-------------")

results = library.search_books("John")

for book in results:
    print(book.get_description())



# Return a book


print("\nRETURNING BOOK")
print("-------------")

student.return_book(book1)



# Total borrowed books


print("\nTOTAL BORROWED BOOKS")
print("--------------------")

print(Member.total_borrowed)



# Library name


print("\nLIBRARY NAME")
print("------------")

print(library.get_library_name())
````
